# Notebook 02 — ETL Pipeline

## Objective

The objective of this notebook is to clean and transform the raw customer churn dataset into a structured dataset suitable for analysis and machine learning.

Tasks include:

- Handling incorrect data types
- Identifying and treating missing values
- Removing unnecessary columns
- Feature engineering
- Exporting the cleaned dataset

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
raw_data_path = Path("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df = pd.read_csv(raw_data_path)

print("Raw dataset loaded successfully.")
```

```python
# Create a copy of the raw dataset
df_clean = df.copy()

print("Working copy created successfully.")
```

```python
print("Original Dataset :", df.shape)
print("Working Dataset  :", df_clean.shape)
```

# 3. Investigate TotalCharges

The `TotalCharges` column should represent the total amount paid by a customer. We first verify its current data type before performing any transformation.

```python
print(df_clean["TotalCharges"].dtype)
```

```python
blank_values = (df_clean["TotalCharges"] == "").sum()

print(f"Blank Values: {blank_values}")
```

```python
blank_spaces = (df_clean["TotalCharges"] == " ").sum()

print(f"Blank Spaces: {blank_spaces}")
```

```python
df_clean[df_clean["TotalCharges"] == " "]
```

```python
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)
```

```python
print(df_clean["TotalCharges"].dtype)
```

```python
df_clean["TotalCharges"].isnull().sum()
```

## Observation

The `TotalCharges` column was initially stored as a string because it contained blank space values.

Using `pd.to_numeric(errors="coerce")`, the column was successfully converted to a numeric type. Blank space values were converted to `NaN`, making them easier to identify and handle during the missing value treatment phase.

# 4. Handle Missing Values

The `TotalCharges` column contained 11 missing values after converting blank strings to `NaN`.

Since these records represent less than 0.2% of the dataset and correspond to customers with no billing history, they are removed from the dataset to maintain data quality.

```python
print(df_clean.isnull().sum())
```

```python
df_clean = df_clean.dropna()
```

```python
print(df_clean.isnull().sum())
```

```python
print(df_clean.shape)
```

## Observation

- The dataset originally contained 7,043 customer records.
- Eleven records had missing values in the `TotalCharges` column.
- These represented customers with no billing history.
- Since the missing percentage was only 0.16%, the records were removed.
- The cleaned dataset now contains **7,032 customer records**.

```python
df_clean["SeniorCitizen"].value_counts()
```

```python
df_clean["SeniorCitizen"] = df_clean["SeniorCitizen"].map({
    0: "No",
    1: "Yes"
})
```

```python
df_clean["SeniorCitizen"].value_counts()
```

```python
df_clean.info()
```

```python
df_clean.describe(include="all")
```

```python
processed_path = Path("../data/processed/customer_churn_clean.csv")

df_clean.to_csv(processed_path, index=False)

print("Clean dataset saved successfully.")
```

# ETL Summary

The ETL pipeline performed the following operations:

- Created a working copy of the raw dataset.
- Converted `TotalCharges` from string to numeric.
- Converted blank values into `NaN`.
- Removed records containing missing values.
- Converted `SeniorCitizen` into meaningful labels.
